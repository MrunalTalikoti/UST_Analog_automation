```python
import pandas as pd
import re

NUM = r"[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?"

def get_val(param, text):
    m = re.search(param + r"\s*=\s*(" + NUM + r")", text)
    if not m:
        raise ValueError(f"Cannot parse {param} from header: {text}")
    return float(m.group(1))

def transform_dataset(csv_path, metric_name, output_csv=None):
    df = pd.read_csv(csv_path)

    rows = []
    cols = list(df.columns)

    # iterate column triplets: a | b | c
    for i in range(0, len(cols) - 2, 3):
        a_col, b_col, c_col = cols[i], cols[i+1], cols[i+2]

        a = get_val("a", a_col)
        b = get_val("b", b_col)
        c = get_val("c", c_col)

        # rows = d sweep index
        d_series = df[a_col]   # real swept values

        for d_val, metric_val in zip(d_series, df[c_col]):
            if pd.notna(d_val) and pd.notna(metric_val):
                rows.append({
                    "a": a,
                    "b": b,
                    "c": c,
                    "d": d_val,          # ✅ real physical d
                    metric_name: metric_val
                })

     

    out_df = pd.DataFrame(rows)

    if output_csv:
        out_df.to_csv(output_csv, index=False)

    return out_df
```

```python
gain_df = transform_dataset(
    r"E:\VIIT\TY\UST Analog Workflow Automation\Project\UST_Analog_automation\data\raw\Updated_Opam\gain27.csv",
    metric_name="gain",
    output_csv=r"E:\VIIT\TY\UST Analog Workflow Automation\Project\UST_Analog_automation\models\Junaid\opam Transformation 5\gain_clean.csv"
)

print(gain_df.head())
print("Rows, Cols:", gain_df.shape)
```
